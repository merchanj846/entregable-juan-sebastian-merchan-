```python
df["Animal type"] = (
    df["Animal type"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["Animal type"].value_counts()
```

```python
df["Animal type"] = (
    df["Animal type"]
    .str.replace("Red Squirrell", "Red Squirrel", regex=False)
    .str.replace("Red Squirel", "Red Squirrel", regex=False)
    .str.replace("European Bison™", "European Bison", regex=False)
    .str.replace("European Bisson", "European Bison", regex=False)
    .str.replace("European Buster", "European Bison", regex=False)
    .str.replace("Ledgehod", "Hedgehog", regex=False)
    .str.replace("Wedgehod", "Hedgehog", regex=False)
    .str.replace("Lynx?", "Lynx", regex=False))
df["Animal type"].value_counts()
```

```python
df["Country"] = (
    df["Country"]
    .str.strip()
    .str.lower()
    .str.title()
)

df["Country"].value_counts()
```

```python
df["Country"]= (
    df["Country"]
    .str.replace("Pl", "Poland", regex=False)
    .str.replace("Hungry", "Hungary", regex=False)
    .str.replace("Hu", "Hungary", regex=False)
    .str.replace("Cz", "Czech Republic", regex=False)
    .str.replace("De", "Germany", regex=False)
)
df.Country.value_counts()
```

```python
df["Country"]= (
    df["Country"]
    .str.replace("Czech Republicech Republic", "Czech Republic", regex=False)
    .str.replace("Czech Republicech", "Czech Republic", regex=False)
    .str.replace("Hungaryngary", "Hungary", regex=False)
)
df.Country.value_counts()
```

```python
df = df.dropna(subset=["Animal type"])
df["Animal type"].isnull().sum()
```

```python
df["Country"].isnull().sum()
```

```python
display(df[df["Weight kg"].isnull()].head(27))
Red_Squirrel_weight = df[df["Animal type"] == "Red Squirrel"]
Hedgehog_weight = df[df["Animal type"] == "Hedgehog"]
df["Weight kg"] = df["Weight kg"].fillna(Red_Squirrel_weight["Weight kg"].median())
df["Weight kg"] = df["Weight kg"].fillna(Hedgehog_weight["Weight kg"].median())
df["Weight kg"].isnull().sum()
```

```python
display(df[df["Body Length cm"].isnull()].head(27))
red_squirrel_median = df[df["Animal type"] == "Red Squirrel"]["Body Length cm"].median()
hedgehog_median = df[df["Animal type"] == "Hedgehog"]["Body Length cm"].median()
df["Body Length cm"] = df["Body Length cm"].fillna(red_squirrel_median)
df["Body Length cm"] = df["Body Length cm"].fillna(hedgehog_median)
df["Body Length cm"].isnull().sum()
```

```python
# Utiliza NumPy para realizar transformaciones sobre el DataFrame. • Crea una variable utilizando: np.where() La variable debe representar una condición relevante para el contexto de los datos. • Utiliza al menos una función estadística de NumPy para calcular una medida sobre una variable numérica. Puedes utilizar, por ejemplo: np.mean() np.median() np.nanmean() np.nanmedian() np.std() Explica brevemente qué representa el resultado obtenido.
df.describe()
```

```python
df['Weight kg'] =df['Weight kg'].abs()
df['Body Length cm'] =df['Body Length cm'].abs()
```

```python
df['Animal type'].value_counts()
```

```python
df['Order @ animal kingdom'] = np.where(
    df['Animal type'] == 'Red Squirrel',
    'Orden Rodentia',
    np.where(
        df['Animal type'] == 'European Bison',
        'Orden Artiodactyla',
        np.where(
            df['Animal type'] == 'Lynx',
            'Orden Carnivora',
            np.where(
                df['Animal type'] == 'Hedgehog',
                'Orden Eulipotyphla',
                'Other'
            )
        )
    )
)
df.head()
```

```python
# creamos una columna utilizando las tres condicionales , indicando cada tipo de animal con su orden , y si no se cumple ninguna poniendo "other" dado que ya lo datos estaban corregidos solamente mirariamos a los animales que ya se encuentran en el data frame
# 4. Análisis con groupby
# Cuál es el número de observaciones por tipo de animal? • Cuál es el promedio de una variable numérica según el tipo de animal? • Cuál es el promedio de una variable numérica según el país?
```

```python
numero_observaciones = df.groupby('Animal type')['Animal type'].count()
print("las observaciones son:")
print(numero_observaciones)
```

```python
# Cuál es el promedio de una variable numérica según el tipo de animal?
peso_promedio= df.groupby('Animal type')['Weight kg'].mean()
print("Promedio de peso por tipo de animal:", peso_promedio)
```

```python
#Cuál es el promedio de una variable numérica según el país? 
peso_por_pais = df.groupby('Country')['Weight kg'].mean()
print("Promedio de peso por país:", peso_por_pais)
```
